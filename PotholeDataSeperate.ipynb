```python
import os

all_file_list=[]
for (path, dir, files) in os.walk("/home/jovyan/data/origin"):
    for filename in files:
        all_file_list.append(path+'/'+filename)
```

```python
len(all_file_list)
```

```python
normal_file_list=[]
pothole_file_list=[]

for i in range(len(all_file_list)):
    file_name=all_file_list[i]
    if  'normal' in file_name:
        normal_file_list.append(file_name)
    else:
        pothole_file_list.append(file_name)
```

```python
from sklearn.model_selection import train_test_split
x_train_normal,x_valid_normal=train_test_split(normal_file_list, test_size=0.2, shuffle=False)
x_test_normal,x_valid_normal=train_test_split(x_valid_normal, test_size=0.5, shuffle=False)
```

```python
len(x_valid_normal)+len(x_test_normal)+len(x_train_normal)
```

```python
x_train_pothole,x_valid_pothole=train_test_split(pothole_file_list, test_size=0.2, shuffle=False)
x_test_pothole,x_valid_pothole=train_test_split(x_valid_pothole, test_size=0.5, shuffle=False)
```

```python
len(x_valid_pothole)+len(x_train_pothole)+len(x_test_pothole)
```

```python
x_test_pothole
```

```python
import shutil
```

```python
train_order=["train","valid","test"]
data_order=["normal","potholes"]
normal_list=[x_train_normal,x_valid_normal,x_test_normal]
pothole_list=[x_train_pothole,x_valid_pothole,x_test_pothole]
```

```python
temp_dic_normal = { name:value for name, value in zip(train_order,normal_list) }
temp_dic_pothole = { name:value for name, value in zip(train_order,pothole_list) }
```

```python
data_split_dic={ name:value for name, value in zip(data_order,[temp_dic_normal,temp_dic_pothole]) }
```

```python
total_sum=0
base_dir="/home/jovyan/data/classification/"
for class_,type_dict in data_split_dic.items():
    for data_type,value_list in type_dict.items():
        print("******",class_,data_type,"******")
        for file_name in value_list:
            print(base_dir+data_type+'/'+class_+'/'+file_name)
            shutil.copy2(file_name,base_dir+data_type+'/'+class_+'/')
```
