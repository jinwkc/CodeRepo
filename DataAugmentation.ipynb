```python
import imgaug.augmenters as iaa
import numpy as np
import cv2
import os
```

```python
data_path     = '/home/jovyan/data/classification/'
aug_data_path = '/home/jovyan/data/classification_AugData/'
```

```python
def load_image_to_path(imagepath_list,image_size=128):

    train_normal=[]
    train_pothole=[]
    valid_normal=[]
    valid_pothole=[]
    test_normal=[]
    test_pothole=[]
    
    train_normal_path=[]
    train_pothole_path=[]
    valid_normal_path=[]
    valid_pothole_path=[]
    test_normal_path=[]
    test_pothole_path=[]
    
    for image in imagepath_list:
        try:
            img=cv2.imread(image,cv2.IMREAD_COLOR)
            img=cv2.resize(img,(image_size,image_size))
        
            if 'train' in image:
                
                if 'normal' in image :
                    train_normal.append(img)
                    train_normal_path.append(image)
                else:
                    train_pothole.append(img)
                    train_pothole_path.append(image)
                    
            elif 'test' in image:
                if 'normal' in image :
                    test_normal.append(img)
                    test_normal_path.append(image)
                else:
                    test_pothole.append(img)
                    test_pothole_path.append(image)
                    
            else:
                if 'normal' in image :
                    valid_normal.append(img)
                    valid_normal_path.append(image)
                else:
                    valid_pothole.append(img)
                    valid_pothole_path.append(image)
        except:
            print(image)
            pass
        
    return train_normal, train_normal_path, train_pothole, train_pothole_path, valid_normal, valid_normal_path, valid_pothole, valid_pothole_path, test_normal, test_normal_path, test_pothole, test_pothole_path
```

```python
def load_data_np_array(data_dir_path,img_size=128):
    imagepath_list=[]
   
    for dirname,_,filenames in os.walk(data_dir_path):
        for filename in filenames:
            path=os.path.join(dirname,filename)
            imagepath_list.append(path)
    
    return load_image_to_path(imagepath_list,image_size=img_size)
```

```python
aug_pipeline = iaa.Sequential([
    iaa.Resize({"height": 128, "width": 128}),  # resize to 256x256
    iaa.Fliplr(0.4),  # horizontal flip with probability 0.5
    iaa.Flipud(0.5),
    iaa.GaussianBlur(sigma=(0, 3.0)),  # add Gaussian blur with sigma between 0 and 3.0
    # iaa.AdditiveGaussianNoise(loc=0, scale=(0.0, 0.05*255), per_channel=0.5),  # add Gaussian noise
    iaa.Multiply((0.5, 1.5), per_channel=0.5),  # change brightness of images (50-150% of original value)
    # iaa.Affine(rotate=(-20, 20), shear=(-5, 5)),  # apply random affine transformations
])
```

```python
for i in range(len(train_normal)):

    for j in range(10):
        aug_img=aug_pipeline(image=train_normal[i])
        file_name=train_normal_path[i].split(sep='/')[-1].split(sep='.')[0]
        if j is 0:
            cv2.imwrite(aug_data_path +'train/normal/{}.jpg'.format(file_name), train_normal[i])
            cv2.imwrite((aug_data_path +'train/normal/{}_aug_{}.jpg'.format(file_name,j)), aug_img)
        else:
            cv2.imwrite((aug_data_path +'train/normal/{}_aug_{}.jpg'.format(file_name,j)),aug_img)
        
        
```
